## Estimation & Confidence Intervals

### Goals
- Understand Point Estimation

- Build an approximate 95% Confidence Interval
- Distinguish Standard Deviation from Standard Error

```python
import numpy as np
import pandas as pd
```

## Example Sample

Suppose these values represent walking distance to the nearest park, in meters.

```python
sample = np.array([
    540, 610, 700, 580, 650,
    620, 590, 670, 630, 610
])

sample
```

## Point Estimate

The sample mean is used as a point estimate of the unknown population mean.

```python
mean = sample.mean()

print("Sample Mean:", mean)
```

## Standard Deviation

Standard Deviation describes the spread of the observations themselves.

```python
std = sample.std(ddof=1)

print("Sample Standard Deviation:", std)
```

## Standard Error

Standard Error describes uncertainty in the estimate of the sample mean.

Formula:

SE = s / sqrt(n)

```python
n = len(sample)

se = std / np.sqrt(n)

print("Sample Size:", n)
print("Standard Error:", se)
```

## Margin of Error

For an approximate 95% Confidence Interval:

Margin of Error ≈ 1.96 × Standard Error

```python
margin = 1.96 * se

print("Margin of Error:", margin)
```

```python
lower = mean - margin
upper = mean + margin

print("Mean:", mean)
print("95% CI:", lower, "to", upper)
```

## Interpretation

The sample mean is our current best estimate of the population mean.

The confidence interval expresses uncertainty around that estimate.

A narrower interval indicates higher precision.

# Standard Deviation vs Standard Error

Standard Deviation:
describes variability among observations.

Standard Error:
describes uncertainty in an estimated statistic.

They answer different questions.

```python
print("Standard Deviation:", std)
print("Standard Error:", se)
```

# Effect of Sample Size

As sample size increases, Standard Error usually decreases.

This generally produces narrower confidence intervals.

```python
np.random.seed(42)

population = np.random.normal(
    loc=600,
    scale=100,
    size=10000
)

population.mean()
```

```python
sample_sizes = [20, 50, 100, 500]

results = []

for n in sample_sizes:

    current_sample = np.random.choice(
        population,
        size=n,
        replace=False
    )

    current_mean = current_sample.mean()
    current_std = current_sample.std(ddof=1)
    current_se = current_std / np.sqrt(n)
    current_margin = 1.96 * current_se

    results.append({
        "sample_size": n,
        "mean": current_mean,
        "standard_error": current_se,
        "ci_lower": current_mean - current_margin,
        "ci_upper": current_mean + current_margin,
        "ci_width": 2 * current_margin
    })

results_df = pd.DataFrame(results)

results_df
```

## Interpretation

As Sample Size increases:

- Standard Error generally decreases
- Confidence Interval width generally decreases
- Precision improves

But larger Sample Size does NOT automatically remove Sampling Bias.

# Precision vs Accuracy

Precision:
How narrow and stable the estimate is.

Accuracy:
How close the estimate is to the true population value.

A result can be highly precise but still inaccurate if the sampling process is biased.

```python
np.random.seed(10)

group_a = np.random.normal(
    loc=500,
    scale=40,
    size=5000
)

group_b = np.random.normal(
    loc=800,
    scale=40,
    size=5000
)

full_population = np.concatenate([
    group_a,
    group_b
])

true_mean = full_population.mean()

biased_sample = np.random.choice(
    group_a,
    size=1000,
    replace=False
)

biased_mean = biased_sample.mean()
biased_std = biased_sample.std(ddof=1)
biased_se = biased_std / np.sqrt(len(biased_sample))

print("True Population Mean:", true_mean)
print("Biased Sample Mean:", biased_mean)
print("Standard Error:", biased_se)
```

## Interpretation

The biased sample may have a very small Standard Error because it is large.

But the estimate is still systematically wrong because it excludes part of the population.

Therefore:

High Precision ≠ High Accuracy

```python
study_a = {
    "mean": 50,
    "lower": 49,
    "upper": 51
}

study_b = {
    "mean": 50,
    "lower": 35,
    "upper": 65
}

width_a = study_a["upper"] - study_a["lower"]
width_b = study_b["upper"] - study_b["lower"]

print("Study A Width:", width_a)
print("Study B Width:", width_b)
```

## Interpretation

Both studies have the same mean.

Study A is more precise because its confidence interval is much narrower.

# Tehran Project Connection

Urban indicators should not always be reported as single numbers.

When indicators are estimated from samples, we should report:

Estimate + Uncertainty

```python
tehran_example = pd.DataFrame({
    "district": ["A", "B", "C"],
    "estimate": [620, 650, 590],
    "ci_lower": [580, 400, 560],
    "ci_upper": [660, 900, 620]
})

tehran_example["ci_width"] = (
    tehran_example["ci_upper"]
    - tehran_example["ci_lower"]
)

tehran_example
```

## Urban Interpretation

District B has the largest point estimate.

However, its confidence interval is also much wider.

Therefore, its estimate is less precise.

For Site Ranking or Urban Intelligence, we should consider uncertainty before making strong decisions.

# Project Rule

Whenever an urban indicator is estimated from a sample, report:

1. Estimate
2. Confidence Interval or another uncertainty measure
3. Sampling Method
4. Sample Size
5. Possible Bias
6. Interpretation Limitations

Do not rank sites based only on a single point estimate.

## E010 — Confidence Interval Interpretation

Incorrect interpretation:

"There is a 95% probability that the true parameter is inside this specific interval."

Better frequentist interpretation:

The procedure used to generate the interval will capture the true population parameter in approximately 95% of repeated samples.

# Session Summary

Key concepts:

- A Point Estimate gives a single estimate of a population parameter.
- Standard Error measures uncertainty in the statistic.
- Margin of Error determines the range around the estimate.
- Confidence Intervals communicate uncertainty.
- Narrower intervals indicate higher precision.
- Precision is not the same as Accuracy.
- Sample Size affects Standard Error.
- Bias cannot be solved only by increasing Sample Size.
- Urban indicators should include uncertainty when possible.
