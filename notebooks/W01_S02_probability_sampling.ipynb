## Example Dataset

Each row represents one hypothetical neighborhood.

Variables:
- district
- land_use
- metro_access

```python
data = {
    "district": ["A","B","C","D","E","F","G","H","I","J"],
    "land_use": [
        "Residential",
        "Commercial",
        "Residential",
        "Administrative",
        "Residential",
        "Commercial",
        "Residential",
        "Administrative",
        "Commercial",
        "Residential"
    ],
    "metro_access": [
        True, True, False, True, False,
        True, False, True, False, True
    ]
}

df = pd.DataFrame(data)

df
```

## Simple Probability

Probability can be estimated from observed proportions.

Example:

P(Metro Access)

```python
p_metro = df["metro_access"].mean()

print("P(Metro Access) =", p_metro)
```

```python
df["land_use"].value_counts()
```

## Probability of an Event

There are 3 Commercial neighborhoods among 10 observations.

Therefore:

P(Commercial) = 3 / 10 = 0.3

```python
p_commercial = (df["land_use"] == "Commercial").mean()

print("P(Commercial) =", p_commercial)
```

## Conditional Probability

We now estimate:

P(Metro Access | Commercial)

Meaning:

Among Commercial neighborhoods, what proportion has metro access?

```python
commercial = df[df["land_use"] == "Commercial"]

commercial
```

```python
p_metro_given_commercial = commercial["metro_access"].mean()

print(
    "P(Metro Access | Commercial) =",
    p_metro_given_commercial
)
```

## Important Interpretation

P(Metro | Commercial) does not mean:

P(Commercial | Metro)

Conditional probability has direction.

```python
metro = df[df["metro_access"] == True]

p_commercial_given_metro = (
    metro["land_use"] == "Commercial"
).mean()

print(
    "P(Commercial | Metro) =",
    p_commercial_given_metro
)
```

## Independence

Two events are independent if:

P(A | B) = P(A)

Equivalently:

P(A ∩ B) = P(A) × P(B)

We should not assume independence without checking.

```python
p_metro = df["metro_access"].mean()

p_metro_given_commercial = (
    df[df["land_use"] == "Commercial"]["metro_access"].mean()
)

print("P(Metro) =", p_metro)
print(
    "P(Metro | Commercial) =",
    p_metro_given_commercial
)
```

### Interpretation

P(Metro) and P(Metro | Commercial) are different in this small dataset.

This suggests that Metro Access and Commercial Land Use are not behaving as independent events in this sample.

This does NOT prove causation.

## Random Variable

A random variable converts uncertain outcomes into numerical values.

For example:

X = number of neighborhoods with Metro Access

```python
metro_count = df["metro_access"].sum()

print(
    "Number of neighborhoods with Metro Access:",
    metro_count
)
```

# Tehran Project Connection

Raw counts can be misleading.

For public-space analysis, a district with more parks may also have:
- more population
- larger area

Therefore we may need normalization.

```python
urban_data = pd.DataFrame({
    "district": ["A", "B", "C"],
    "population": [50000, 100000, 75000],
    "area_km2": [5, 8, 6],
    "parks": [10, 16, 12]
})

urban_data
```

```python
urban_data["parks_per_10000_people"] = (
    urban_data["parks"]
    / urban_data["population"]
    * 10000
)

urban_data["parks_per_km2"] = (
    urban_data["parks"]
    / urban_data["area_km2"]
)

urban_data
```

## Interpretation

Raw park count answers:

"How many parks are there?"

Parks per population answers:

"How much park availability exists relative to residents?"

Parks per area answers:

"How spatially dense are parks?"

These are different urban questions.

# Session Summary

Key concepts:
